# 09 — Temporal Fusion Transformer (TFT) for Groundwater Level Forecasting

**Goal:** Multi-horizon prediction of district-level groundwater levels (t+7, t+14, t+30)  
**Output:** q10 / q50 / q90 quantile forecasts per district per horizon  
**Reference:** Lim et al. (2021) — *Temporal Fusion Transformers for Interpretable Multi-horizon Time Series Forecasting*

---
### Pipeline
1. Load & impute `master_dataset.csv`
2. Normalise features
3. Build sliding-window dataset
4. Train TFT with quantile loss + early stopping
5. Evaluate at t+7, t+14, t+30
6. Quantile interval plots
7. VSN feature importance
8. Attention heatmaps
9. Save predictions + model

```python
import sys, warnings, json, time
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
import seaborn as sns
from sklearn.preprocessing import StandardScaler

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

sys.path.insert(0, str(Path('.').resolve()))
from tft_model import TemporalFusionTransformer, QuantileLoss

warnings.filterwarnings('ignore')
sys.stdout.reconfigure(encoding='utf-8', errors='replace')

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Device:', DEVICE)
if DEVICE.type == 'cuda':
    print('GPU:', torch.cuda.get_device_name(0))
```

```python
# ─── CONFIG ──────────────────────────────────────────────────────────────────
ENCODER_LEN    = 90          # past days of context fed to LSTM encoder
DECODER_LEN    = 30          # future days predicted (evaluate at steps 7, 14, 30)
EVAL_STEPS     = [6, 13, 29] # 0-indexed: step 6 = t+7, 13 = t+14, 29 = t+30

D_MODEL        = 128         # hidden dimension
N_HEADS        = 4           # attention heads
LSTM_LAYERS    = 2
DROPOUT        = 0.1
QUANTILES      = [0.1, 0.5, 0.9]
CAT_EMBED_DIM  = 8           # district embedding size

BATCH_SIZE     = 64
MAX_EPOCHS     = 100
LR             = 1e-3
PATIENCE       = 15          # early stopping patience
GRAD_CLIP      = 1.0

TRAIN_END      = '2025-01-01'  # train up to here
# val = everything after TRAIN_END (last ~3.5 months)

SEED           = 42
torch.manual_seed(SEED)
np.random.seed(SEED)

OUT_DIR = Path('outputs')
OUT_DIR.mkdir(exist_ok=True)

DATA_PATH = Path('../../Data Preprocessing/clean datasets/master_dataset.csv')
```

## 1. Load & Impute Data

```python
df = pd.read_csv(DATA_PATH, parse_dates=['date'])
df = df.sort_values(['district', 'date']).reset_index(drop=True)
print(f'Raw shape: {df.shape}')
print(f'Districts: {df["district"].nunique()}')
print(f'Date range: {df["date"].min().date()} → {df["date"].max().date()}')

# ── impute soil texture for 4 districts with no soil data ────────────────────
SOIL_COLS = [
    'sand_0cm','clay_0cm','silt_0cm',
    'sand_10cm','clay_10cm','silt_10cm',
    'sand_30cm','clay_30cm','silt_30cm'
]
soil_medians = df[SOIL_COLS].median()
df[SOIL_COLS] = df[SOIL_COLS].fillna(soil_medians)
print('Soil imputed with dataset medians')

# ── impute Sentinel for 4 districts with no coverage ─────────────────────────
SENTINEL_COLS = ['ndvi_mean','ndwi_mean','ndmi_mean','vv_db','vh_db','vh_vv_ratio']
sent_medians = df[SENTINEL_COLS].median()
df[SENTINEL_COLS] = df[SENTINEL_COLS].fillna(sent_medians)
print('Sentinel imputed with dataset medians')

# ── fill lag NaNs at start of each district (first 30 rows have no 30-day lag) ─
LAG_COLS = ['gwl_lag_7','gwl_lag_14','gwl_lag_30','gwl_7d_change','gwl_30d_change']
df[LAG_COLS] = df.groupby('district')[LAG_COLS].transform(lambda s: s.fillna(s.median()))

# ── encode season as integer ──────────────────────────────────────────────────
SEASON_MAP = {'winter': 0, 'summer': 1, 'monsoon': 2, 'post_monsoon': 3}
df['season_id'] = df['season'].map(SEASON_MAP).astype(int)

print(f'\nMissing after imputation:')
print(df.isnull().sum()[df.isnull().sum() > 0])
print(f'\nTotal rows: {len(df):,}')
```

## 2. Feature Definitions

```python
# ── feature groups ────────────────────────────────────────────────────────────

# static categorical: district_id (will be embedded)
STATIC_CAT = ['district_id']

# static continuous: soil texture (never changes per district)
STATIC_CONT = SOIL_COLS  # 9 features

# observed past: everything known only up to today
PAST_CONT = [
    'gwl_mean',
    'gwl_min', 'gwl_max', 'station_count',
    'rainfall_mm', 'rainfall_7d', 'rainfall_30d',
    'et_mm', 'soil_moisture', 'net_water_balance',
    'temp_max', 'temp_min', 'temp_mean', 'humidity_mean', 'vpd_approx',
    'gwl_lag_7', 'gwl_lag_14', 'gwl_lag_30',
    'gwl_7d_change', 'gwl_30d_change',
    'ndvi_mean', 'ndwi_mean', 'ndmi_mean',
    'vv_db', 'vh_db', 'vh_vv_ratio',
]  # 26 features

# known future: calendar features you can compute for any future date
FUTURE_CONT = ['month', 'day_of_year', 'is_monsoon', 'season_id']  # 4 features

TARGET = 'gwl_mean'

print(f'Past features    : {len(PAST_CONT)}')
print(f'Future features  : {len(FUTURE_CONT)}')
print(f'Static cont      : {len(STATIC_CONT)}')
print(f'Static cat       : {len(STATIC_CAT)}')

N_DISTRICTS = df['district_id'].nunique()
print(f'\nDistricts: {N_DISTRICTS}')
print(f'District ID range: {df["district_id"].min()} – {df["district_id"].max()}')
```

## 3. Normalisation

```python
# fit scaler on TRAINING data only
train_mask = df['date'] < TRAIN_END

cols_to_scale = PAST_CONT + STATIC_CONT + FUTURE_CONT
scaler = StandardScaler()
scaler.fit(df.loc[train_mask, cols_to_scale])

df_norm = df.copy()
df_norm[cols_to_scale] = scaler.transform(df[cols_to_scale])

# separate target scaler for inverse-transforming predictions
target_scaler = StandardScaler()
target_scaler.fit(df.loc[train_mask, [TARGET]])
df_norm[TARGET] = target_scaler.transform(df[[TARGET]])

print('Normalised. Target mean (train):', df.loc[train_mask, TARGET].mean().round(3))
print('Target std  (train):', df.loc[train_mask, TARGET].std().round(3))
```

## 4. Dataset & DataLoader

```python
class GWLDataset(Dataset):
    """
    Sliding window dataset for TFT.

    For each valid (district, date) anchor:
        - past window  : [anchor - encoder_len : anchor]
        - future window: [anchor : anchor + decoder_len]
        - target       : gwl_mean at future steps

    Static features are taken from the anchor row (same value for whole district).
    """

    def __init__(self, df, encoder_len, decoder_len,
                 past_cols, future_cols, static_cat_cols, static_cont_cols, target_col,
                 split='train', train_end=TRAIN_END):

        self.enc  = encoder_len
        self.dec  = decoder_len
        self.windows = []

        for dist, grp in df.groupby('district'):
            grp = grp.sort_values('date').reset_index(drop=True)
            n   = len(grp)

            for i in range(encoder_len, n - decoder_len + 1):
                anchor_date = grp.loc[i - 1, 'date']

                # split filter
                if split == 'train' and anchor_date >= pd.Timestamp(train_end):
                    continue
                if split == 'val' and anchor_date < pd.Timestamp(train_end):
                    continue

                past_df   = grp.iloc[i - encoder_len : i]
                future_df = grp.iloc[i : i + decoder_len]

                past_cont   = past_df[past_cols].values.astype(np.float32)
                future_cont = future_df[future_cols].values.astype(np.float32)
                static_cat  = grp.loc[0, static_cat_cols].values.astype(np.int64)
                static_cont = grp.loc[0, static_cont_cols].values.astype(np.float32)
                target      = future_df[target_col].values.astype(np.float32)

                # skip windows with NaN
                if (np.isnan(past_cont).any() or
                    np.isnan(future_cont).any() or
                    np.isnan(target).any()):
                    continue

                self.windows.append({
                    'past_cont':   past_cont,
                    'future_cont': future_cont,
                    'static_cat':  static_cat,
                    'static_cont': static_cont,
                    'target':      target,
                    'district':    dist,
                    'date':        future_df['date'].values[-1],
                })

    def __len__(self):
        return len(self.windows)

    def __getitem__(self, idx):
        w = self.windows[idx]
        return (
            torch.from_numpy(w['static_cat']),
            torch.from_numpy(w['static_cont']),
            torch.from_numpy(w['past_cont']),
            torch.from_numpy(w['future_cont']),
            torch.from_numpy(w['target']),
        )


print('Building datasets...')
t0 = time.time()
train_ds = GWLDataset(df_norm, ENCODER_LEN, DECODER_LEN,
                      PAST_CONT, FUTURE_CONT, STATIC_CAT, STATIC_CONT, TARGET,
                      split='train')
val_ds   = GWLDataset(df_norm, ENCODER_LEN, DECODER_LEN,
                      PAST_CONT, FUTURE_CONT, STATIC_CAT, STATIC_CONT, TARGET,
                      split='val')
print(f'Done in {time.time()-t0:.1f}s')
print(f'Train windows: {len(train_ds):,}')
print(f'Val windows  : {len(val_ds):,}')

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,  num_workers=0, pin_memory=True)
val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False, num_workers=0, pin_memory=True)
```

## 5. Model Initialisation

```python
model = TemporalFusionTransformer(
    n_static_cat    = len(STATIC_CAT),
    cat_vocab_sizes = [N_DISTRICTS],
    cat_embed_dim   = CAT_EMBED_DIM,
    n_static_cont   = len(STATIC_CONT),
    n_past          = len(PAST_CONT),
    n_future        = len(FUTURE_CONT),
    encoder_len     = ENCODER_LEN,
    decoder_len     = DECODER_LEN,
    d_model         = D_MODEL,
    n_heads         = N_HEADS,
    lstm_layers     = LSTM_LAYERS,
    dropout         = DROPOUT,
    quantiles       = QUANTILES,
).to(DEVICE)

criterion = QuantileLoss(QUANTILES).to(DEVICE)
optimizer = torch.optim.Adam(model.parameters(), lr=LR, weight_decay=1e-5)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode='min', factor=0.5, patience=5, min_lr=1e-6
)

n_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f'TFT parameters: {n_params:,}')
print(model)
```

## 6. Training Loop

```python
def run_epoch(loader, model, criterion, optimizer=None, grad_clip=None):
    """One forward pass over loader. If optimizer is None → eval mode."""
    training = optimizer is not None
    model.train(training)
    total_loss, n_batches = 0.0, 0

    with torch.set_grad_enabled(training):
        for batch in loader:
            sc, sx, px, fx, y = [b.to(DEVICE) for b in batch]
            preds, _, _ = model(sc, sx, px, fx)  # (B, T_dec, n_q)
            loss = criterion(preds, y)

            if training:
                optimizer.zero_grad()
                loss.backward()
                if grad_clip:
                    nn.utils.clip_grad_norm_(model.parameters(), grad_clip)
                optimizer.step()

            total_loss += loss.item()
            n_batches  += 1

    return total_loss / n_batches


train_losses, val_losses = [], []
best_val_loss = float('inf')
patience_counter = 0
best_state = None

print(f'Training TFT for up to {MAX_EPOCHS} epochs (early stopping patience={PATIENCE})')
print(f'Train batches/epoch: {len(train_loader)}')
print('-' * 60)

for epoch in range(1, MAX_EPOCHS + 1):
    t0 = time.time()

    train_loss = run_epoch(train_loader, model, criterion, optimizer, GRAD_CLIP)
    val_loss   = run_epoch(val_loader,   model, criterion)
    scheduler.step(val_loss)

    train_losses.append(train_loss)
    val_losses.append(val_loss)

    elapsed = time.time() - t0
    lr_now  = optimizer.param_groups[0]['lr']
    print(f'Epoch {epoch:03d} | train {train_loss:.4f} | val {val_loss:.4f} | lr {lr_now:.2e} | {elapsed:.1f}s')

    if val_loss < best_val_loss:
        best_val_loss    = val_loss
        patience_counter = 0
        best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}
    else:
        patience_counter += 1
        if patience_counter >= PATIENCE:
            print(f'\nEarly stopping at epoch {epoch} (best val loss: {best_val_loss:.4f})')
            break

# restore best weights
model.load_state_dict(best_state)
torch.save(best_state, OUT_DIR / 'tft_checkpoint.pt')
print('\nBest model saved → outputs/tft_checkpoint.pt')
```

```python
# ── Training curves ───────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(train_losses, label='Train loss')
ax.plot(val_losses,   label='Val loss')
ax.set_xlabel('Epoch')
ax.set_ylabel('Quantile loss')
ax.set_title('TFT Training Curve')
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(OUT_DIR / 'tft_training_curve.png', dpi=150)
plt.show()
print('Saved → outputs/tft_training_curve.png')
```

## 7. Evaluation (RMSE / MAE / MAPE at t+7, t+14, t+30)

```python
def evaluate(loader, model, target_scaler, eval_steps, dataset):
    """
    Run inference on a DataLoader and collect per-step predictions.
    Returns a DataFrame with actual / q10 / q50 / q90 columns.
    """
    model.eval()
    records = []

    with torch.no_grad():
        for batch_idx, batch in enumerate(loader):
            sc, sx, px, fx, y = [b.to(DEVICE) for b in batch]
            preds, _, _ = model(sc, sx, px, fx)  # (B, T_dec, 3)

            # inverse transform
            B = y.shape[0]
            for step in eval_steps:
                horizon = step + 1  # step 6 → horizon 7
                act = target_scaler.inverse_transform(
                    y[:, step].cpu().numpy().reshape(-1, 1)).flatten()
                q10 = target_scaler.inverse_transform(
                    preds[:, step, 0].cpu().numpy().reshape(-1, 1)).flatten()
                q50 = target_scaler.inverse_transform(
                    preds[:, step, 1].cpu().numpy().reshape(-1, 1)).flatten()
                q90 = target_scaler.inverse_transform(
                    preds[:, step, 2].cpu().numpy().reshape(-1, 1)).flatten()

                start = batch_idx * BATCH_SIZE
                for i in range(B):
                    w = dataset.windows[start + i]
                    records.append({
                        'district': w['district'],
                        'date':     str(w['date']),
                        'horizon':  f't+{horizon}',
                        'actual':   float(act[i]),
                        'q10':      float(q10[i]),
                        'q50':      float(q50[i]),
                        'q90':      float(q90[i]),
                    })

    return pd.DataFrame(records)


def compute_metrics(df_pred):
    """Compute RMSE, MAE, MAPE for each horizon."""
    rows = []
    for h, grp in df_pred.groupby('horizon'):
        act, pred = grp['actual'].values, grp['q50'].values
        rmse = np.sqrt(np.mean((act - pred) ** 2))
        mae  = np.mean(np.abs(act - pred))
        mape = np.mean(np.abs((act - pred) / (np.abs(act) + 1e-8))) * 100
        rows.append({'Horizon': h, 'RMSE (m)': round(rmse, 4),
                     'MAE (m)': round(mae, 4), 'MAPE (%)': round(mape, 2)})
    return pd.DataFrame(rows).set_index('Horizon')


print('Running evaluation on val set...')
val_preds = evaluate(val_loader, model, target_scaler, EVAL_STEPS, val_ds)
val_metrics = compute_metrics(val_preds)

print('\n=== TFT Val Metrics (q50 = median forecast) ===')
print(val_metrics.to_string())

val_preds.to_csv(OUT_DIR / 'tft_predictions.csv', index=False)
val_metrics.to_csv(OUT_DIR / 'tft_metrics.csv')
print('\nSaved → outputs/tft_predictions.csv')
```

```python
# ── Per-district metrics at t+30 ─────────────────────────────────────────────
t30 = val_preds[val_preds['horizon'] == 't+30']
dist_metrics = []
for dist, grp in t30.groupby('district'):
    act, pred = grp['actual'].values, grp['q50'].values
    rmse = np.sqrt(np.mean((act - pred) ** 2))
    mae  = np.mean(np.abs(act - pred))
    mape = np.mean(np.abs((act - pred) / (np.abs(act) + 1e-8))) * 100
    dist_metrics.append({'District': dist, 'RMSE': round(rmse,3),
                         'MAE': round(mae,3), 'MAPE': round(mape,2)})

dist_df = pd.DataFrame(dist_metrics).sort_values('MAPE')
print('\n=== Per-district MAPE at t+30 ===')
print(dist_df.to_string(index=False))
```

## 8. Quantile Interval Plots

```python
def plot_quantile_forecast(df_pred, district, horizon='t+30', n_days=90):
    """Plot actual vs q10/q50/q90 for a given district and horizon."""
    sub = df_pred[
        (df_pred['district'] == district) &
        (df_pred['horizon']  == horizon)
    ].sort_values('date').tail(n_days)

    if sub.empty:
        print(f'No data for {district} @ {horizon}')
        return

    fig, ax = plt.subplots(figsize=(14, 5))
    x = range(len(sub))

    ax.fill_between(x, sub['q10'], sub['q90'], alpha=0.2, color='steelblue', label='80% interval (q10–q90)')
    ax.plot(x, sub['q50'],    color='steelblue', lw=1.5, label='q50 (median forecast)')
    ax.plot(x, sub['actual'], color='black',    lw=1.5, ls='--', label='Actual GWL')

    ax.set_title(f'{district} — {horizon} quantile forecast')
    ax.set_xlabel('Day')
    ax.set_ylabel('GWL (m, negative = below ground)')
    ax.legend()
    ax.grid(alpha=0.3)
    plt.tight_layout()
    fname = OUT_DIR / f'quantile_{district.lower().replace(" ","_")}_{horizon}.png'
    plt.savefig(fname, dpi=150)
    plt.show()


# Plot for 4 representative districts
sample_districts = val_preds['district'].unique()[:4].tolist()
for dist in sample_districts:
    plot_quantile_forecast(val_preds, dist, horizon='t+30')
```

## 9. VSN Feature Importance

```python
def collect_vsn_weights(loader, model, n_batches=30):
    """Average VSN selection weights over n_batches."""
    model.eval()
    past_w_list, fut_w_list, stat_w_list = [], [], []

    with torch.no_grad():
        for i, batch in enumerate(loader):
            if i >= n_batches:
                break
            sc, sx, px, fx, _ = [b.to(DEVICE) for b in batch]
            _, vsn_w, _ = model(sc, sx, px, fx)
            past_w_list.append(vsn_w['past'].cpu().mean(dim=(0, 1)))   # avg over B and T
            fut_w_list.append(vsn_w['future'].cpu().mean(dim=(0, 1)))
            stat_w_list.append(vsn_w['static'].cpu().mean(dim=0))

    return (
        torch.stack(past_w_list).mean(0).numpy(),
        torch.stack(fut_w_list).mean(0).numpy(),
        torch.stack(stat_w_list).mean(0).numpy(),
    )


past_imp, fut_imp, stat_imp = collect_vsn_weights(val_loader, model)

fig, axes = plt.subplots(1, 3, figsize=(20, 6))

# Past features
ax = axes[0]
idx = np.argsort(past_imp)[::-1]
ax.barh([PAST_CONT[i] for i in idx], past_imp[idx], color='steelblue')
ax.set_title('Past Feature Importance (VSN)')
ax.set_xlabel('Mean VSN weight')
ax.invert_yaxis()

# Future features
ax = axes[1]
idx = np.argsort(fut_imp)[::-1]
ax.barh([FUTURE_CONT[i] for i in idx], fut_imp[idx], color='coral')
ax.set_title('Future Feature Importance (VSN)')
ax.set_xlabel('Mean VSN weight')
ax.invert_yaxis()

# Static features
ax = axes[2]
static_names = ['district_emb'] + STATIC_CONT
idx = np.argsort(stat_imp)[::-1]
ax.barh([static_names[i] for i in idx], stat_imp[idx], color='green')
ax.set_title('Static Feature Importance (VSN)')
ax.set_xlabel('Mean VSN weight')
ax.invert_yaxis()

plt.tight_layout()
plt.savefig(OUT_DIR / 'tft_vsn_importance.png', dpi=150)
plt.show()
print('Saved → outputs/tft_vsn_importance.png')
```

## 10. Attention Heatmap

```python
def plot_attention_heatmap(loader, model, n_samples=5):
    """Show temporal attention weights for a few samples."""
    model.eval()
    batch = next(iter(loader))
    sc, sx, px, fx, y = [b[:n_samples].to(DEVICE) for b in batch]

    with torch.no_grad():
        _, _, attn_w = model(sc, sx, px, fx)
    # attn_w: (B, T_dec, T_enc+T_dec)

    # average across decoder steps → which encoder day was attended to most
    mean_attn = attn_w.mean(dim=1).cpu().numpy()  # (n_samples, T_enc+T_dec)

    fig, axes = plt.subplots(n_samples, 1, figsize=(14, 3 * n_samples))
    for i in range(n_samples):
        ax = axes[i] if n_samples > 1 else axes
        ax.bar(range(mean_attn.shape[1]), mean_attn[i], width=1.0, color='steelblue', alpha=0.7)
        ax.axvline(x=ENCODER_LEN - 0.5, color='red', ls='--', label='Past | Future')
        ax.set_ylabel('Attn weight')
        ax.set_title(f'Sample {i+1} — Temporal Attention')
        if i == 0:
            ax.legend()

    axes[-1].set_xlabel(f'Timestep (0–{ENCODER_LEN-1}: past, {ENCODER_LEN}–{ENCODER_LEN+DECODER_LEN-1}: future)')
    plt.tight_layout()
    plt.savefig(OUT_DIR / 'tft_attention_heatmap.png', dpi=150)
    plt.show()
    print('Saved → outputs/tft_attention_heatmap.png')


plot_attention_heatmap(val_loader, model)
```

## 11. Comparison with Baselines

```python
baseline_path = Path('../baselines/outputs/multihorizon_summary.csv')

if baseline_path.exists():
    baseline = pd.read_csv(baseline_path)
    # Add TFT row
    tft_rows = []
    for h in ['t+7', 't+14', 't+30']:
        row = val_metrics.loc[h] if h in val_metrics.index else None
        if row is not None:
            tft_rows.append({'Model': 'TFT', 'Horizon': h,
                             'RMSE': row['RMSE (m)'],
                             'MAE':  row['MAE (m)'],
                             'MAPE': row['MAPE (%)']})

    baseline_renamed = baseline.rename(columns={'RMSE': 'RMSE', 'MAE': 'MAE', 'MAPE': 'MAPE'})
    combined = pd.concat([baseline_renamed, pd.DataFrame(tft_rows)], ignore_index=True)
    combined.to_csv(OUT_DIR / 'comparison_with_tft.csv', index=False)

    # Pretty table
    pivot = combined.pivot_table(index='Model', columns='Horizon', values='MAPE')
    print('\n=== MAPE (%) Comparison — All Models ===')
    print(pivot.round(2).to_string())
else:
    print('Baseline summary not found — run 08_evaluate_multihorizon.ipynb first')
    print('\nTFT standalone metrics:')
    print(val_metrics.to_string())
```

## 12. Save Summary

```python
summary = {
    'model': 'TFT',
    'encoder_len': ENCODER_LEN,
    'decoder_len': DECODER_LEN,
    'd_model': D_MODEL,
    'n_heads': N_HEADS,
    'lstm_layers': LSTM_LAYERS,
    'dropout': DROPOUT,
    'quantiles': QUANTILES,
    'n_params': n_params,
    'best_val_loss': round(best_val_loss, 6),
    'epochs_trained': len(train_losses),
    'metrics': val_metrics.reset_index().to_dict('records'),
}

with open(OUT_DIR / 'tft_summary.json', 'w') as f:
    json.dump(summary, f, indent=2)

print('=== Final TFT Summary ===')
for k, v in summary.items():
    if k != 'metrics':
        print(f'  {k}: {v}')
print()
print(val_metrics.to_string())
print('\nAll outputs saved to outputs/')
```
