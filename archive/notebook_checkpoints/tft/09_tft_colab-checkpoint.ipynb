# TFT v3 — Groundwater Forecasting, Walk-Forward + Masked Loss (Google Colab)

**What changed from v2:**
- `master_dataset.csv` is now a **full daily calendar** per district (no dropped rows) — every 150-day
  window is genuinely 150 consecutive days. v2 silently trained on windows that spanned months
  because gaps were dropped before windowing.
- `gwl_observed` (0/1) marks which days are real CGWB readings vs. gap-filled. The loss is
  **masked** on unobserved target days instead of pretending they're real.
- Evaluation is now **walk-forward**: 3 folds (monsoon / post-monsoon / dry season), each trained
  on everything before it and tested on a season the model never saw during training or early
  stopping. v2 reported metrics on a single dry-season validation split that was also used for
  early stopping — optimistic and monsoon-blind.
- Districts with <5 stations (Jalgaon, Akola) are flagged (`low_station_flag`) and reported
  separately — their "district mean" is 1-2 wells, not a district signal.
- Inverse-frequency sample weighting (capped 3x) so low-coverage districts (Dhule, Beed, etc.)
  aren't drowned out by high-coverage ones during training.

```python
import sys, warnings, json, time, math
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# NOTE: import torch BEFORE sklearn. On some environments (observed on Windows,
# harmless on Colab but kept for portability/local debugging) importing sklearn
# first and then creating a torch optimizer can segfault due to a BLAS/OpenMP
# DLL conflict between the two libraries' bundled linear algebra backends.
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler

from sklearn.preprocessing import StandardScaler

from tft_model import TemporalFusionTransformer

warnings.filterwarnings('ignore')

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Device:', DEVICE)
if DEVICE.type == 'cuda':
    print('GPU:', torch.cuda.get_device_name(0))
else:
    print('WARNING: No GPU found. Go to Runtime -> Change runtime type -> T4 GPU')
```

```python
# ─── CONFIG ──────────────────────────────────────────────────────────────────
ENCODER_LEN   = 120        # try 90 as an ablation — fixed 843-day span means longer
                           # encoders leave fewer windows per district (esp. Dhule/Beed)
DECODER_LEN   = 30
EVAL_STEPS    = [6, 13, 29]
D_MODEL       = 192
N_HEADS       = 8
LSTM_LAYERS   = 3
DROPOUT       = 0.15
QUANTILES     = [0.1, 0.5, 0.9]
CAT_EMBED_DIM = 16
BATCH_SIZE    = 128
MAX_EPOCHS    = 120        # per fold
LR            = 1e-3
WARMUP_EPOCHS = 5
PATIENCE      = 20
GRAD_CLIP     = 1.0
SEED          = 42
Q50_LOSS_BOOST= 1.0

# Window-quality gate: a window must have this fraction of REAL (gwl_observed=1)
# days to be used. Encoder gate is stricter (it's the model's evidence); decoder
# gate is looser (some missing target days inside a window are fine — masked, not dropped).
ENC_OBS_MIN = 0.70
DEC_OBS_MIN = 0.50

# Sample weighting: inverse per-district train-window count, capped at this ratio
# vs. the best-covered district, so Nagpur (550 windows) doesn't drown out
# Dhule (168 windows) purely by volume.
SAMPLE_WEIGHT_CAP = 3.0

# Walk-forward folds — each trains on everything before it, tests on a full season
# it has never seen (not even for early stopping). Early-stop validation = the
# last VAL_MONTHS of the fold's own training span.
VAL_MONTHS = 2
FOLDS = [
    {"name": "fold1_monsoon",     "test_start": "2024-07-01", "test_end": "2024-09-30"},
    {"name": "fold2_postmonsoon", "test_start": "2024-10-01", "test_end": "2024-12-31"},
    {"name": "fold3_dry",         "test_start": "2025-01-01", "test_end": "2025-04-22"},
]

torch.manual_seed(SEED)
np.random.seed(SEED)

OUT_DIR   = Path('outputs')
OUT_DIR.mkdir(exist_ok=True)
DATA_PATH = Path('master_dataset.csv')

print(f'Config: d_model={D_MODEL}, encoder={ENCODER_LEN}, lstm_layers={LSTM_LAYERS}, heads={N_HEADS}')
print(f'Folds: {[f["name"] for f in FOLDS]}')
```

```python
# ── Load & impute ─────────────────────────────────────────────────────────────
df = pd.read_csv(DATA_PATH, parse_dates=['date'])
df = df.sort_values(['district','date']).reset_index(drop=True)
print(f'Shape: {df.shape}  |  Districts: {df["district"].nunique()}  |  {df["date"].min().date()} -> {df["date"].max().date()}')
print(f'Overall gwl_observed rate: {df["gwl_observed"].mean():.1%}')

SOIL_COLS     = ['sand_0cm','clay_0cm','silt_0cm','sand_10cm','clay_10cm','silt_10cm','sand_30cm','clay_30cm','silt_30cm']
SENTINEL_COLS = ['ndvi_mean','ndwi_mean','ndmi_mean','vv_db','vh_db','vh_vv_ratio']
CGWB_COLS     = ['annual_recharge_bcm','annual_extraction_bcm','extractable_resource_bcm','stage_of_extraction_pct']

# Safety nets — master_dataset.csv should already be 0-null in these columns
# (the district gate + fixed GAUL name bugs removed the old gaps); kept as a
# defensive fallback so a future re-collection with partial coverage can't crash.
df[SOIL_COLS]     = df[SOIL_COLS].fillna(df[SOIL_COLS].median())
df[SENTINEL_COLS] = df[SENTINEL_COLS].fillna(df[SENTINEL_COLS].median())
df[CGWB_COLS]     = df[CGWB_COLS].fillna(df[CGWB_COLS].median())
df['aquifer_category_id'] = df['aquifer_category_id'].fillna(0).astype(int)
df['season_id'] = df['season'].map({'winter':0,'summer':1,'monsoon':2,'post_monsoon':3}).astype(int)

# ── Split GWL into a RAW target (true readings, NaN preserved) and an
# INTERPOLATED encoder feature (NaN-free, needed because the LSTM can't
# consume NaN inputs). The raw column + gwl_observed mask are what the loss
# and evaluation use; the interpolated column is what the encoder sees.
df['gwl_mean_raw'] = df['gwl_mean'].copy()
for col in ['gwl_mean', 'gwl_min', 'gwl_max']:
    df[col] = df.groupby('district')[col].transform(
        lambda s: s.interpolate('linear', limit_direction='both'))

# Date-true lags/changes recomputed from the interpolated series (master's
# calendar is gap-free per district, verified at build time — see
# preprocess/merge_master.py's contiguity assertion).
df = df.sort_values(['district', 'date'])
df['gwl_lag_7']  = df.groupby('district')['gwl_mean'].shift(7)
df['gwl_lag_14'] = df.groupby('district')['gwl_mean'].shift(14)
df['gwl_lag_30'] = df.groupby('district')['gwl_mean'].shift(30)
df['gwl_7d_change']  = df['gwl_mean'] - df['gwl_lag_7']
df['gwl_30d_change'] = df['gwl_mean'] - df['gwl_lag_30']

print('Missing after imputation (gwl_mean_raw excluded — its NaN pattern is the')
print('true gwl_observed gaps and is expected; lag/change cols below should only')
print('show NaN in the first 7/14/30 rows per district, well before any window'
      ' starts at row ENCODER_LEN):')
check_cols = [c for c in df.columns if c != 'gwl_mean_raw']
print(df[check_cols].isnull().sum().pipe(lambda s: s[s > 0]).to_string() or '  (none)')
```

```python
# ── Feature engineering: cyclic time, log rainfall, rolling gwl stats ─────────
# Cyclic encoding — month/day_of_year are circular, raw integers confuse the model
df['month_sin'] = np.sin(2*np.pi * df['month']       / 12.0)
df['month_cos'] = np.cos(2*np.pi * df['month']       / 12.0)
df['doy_sin']   = np.sin(2*np.pi * df['day_of_year'] / 365.25)
df['doy_cos']   = np.cos(2*np.pi * df['day_of_year'] / 365.25)

# Log-transform heavy-tailed rainfall (many zeros, occasional huge values)
for c in ['rainfall_mm','rainfall_7d','rainfall_30d']:
    df[c+'_log'] = np.log1p(df[c].clip(lower=0))

# Rolling GWL statistics per district, computed on the INTERPOLATED series —
# captures local volatility without inheriting NaN from unobserved days.
for w in [30, 60, 90]:
    df[f'gwl_roll_mean_{w}'] = df.groupby('district')['gwl_mean'].transform(
        lambda s: s.shift(1).rolling(w, min_periods=5).mean())
    df[f'gwl_roll_std_{w}']  = df.groupby('district')['gwl_mean'].transform(
        lambda s: s.shift(1).rolling(w, min_periods=5).std())

# Any remaining NaN in rolling stats only occurs in the first ~30-90 rows of
# each district's calendar (before enough history exists) — well before the
# first possible window anchor at row ENCODER_LEN, so this is a pure safety net.
roll_cols = [c for c in df.columns if c.startswith('gwl_roll_')]
df[roll_cols] = df.groupby('district')[roll_cols].transform(lambda s: s.fillna(s.median()))
df[roll_cols] = df[roll_cols].fillna(0)

print('Engineered features: 4 cyclic + 3 log_rainfall + 6 rolling_gwl =', 4+3+6, 'new cols')
print('Total cols:', df.shape[1])
```

```python
# ── Feature definitions ───────────────────────────────────────────────────────
STATIC_CAT  = ['district_id', 'aquifer_category_id']
STATIC_CONT = SOIL_COLS + [
    'annual_recharge_bcm', 'annual_extraction_bcm',
    'extractable_resource_bcm', 'stage_of_extraction_pct',
]
PAST_CONT   = [
    'gwl_mean','gwl_min','gwl_max','station_count',
    'gwl_observed',   # NEW — 0/1 flag: is this encoder day a real reading or gap-filled?
    'rainfall_mm_log','rainfall_7d_log','rainfall_30d_log',
    'et_mm','soil_moisture','net_water_balance',
    'temp_max','temp_min','temp_mean','humidity_mean','vpd_approx',
    'gwl_lag_7','gwl_lag_14','gwl_lag_30','gwl_7d_change','gwl_30d_change',
    'gwl_roll_mean_30','gwl_roll_std_30','gwl_roll_mean_60','gwl_roll_std_60',
    'gwl_roll_mean_90','gwl_roll_std_90',
    'ndvi_mean','ndwi_mean','ndmi_mean','vv_db','vh_db','vh_vv_ratio',
]
FUTURE_CONT = ['month_sin','month_cos','doy_sin','doy_cos','is_monsoon','season_id']

# TARGET is the RAW (uninterpolated) reading — NaN where gwl_observed==0.
# It is never used as a model input, only as the loss/eval ground truth.
TARGET      = 'gwl_mean_raw'
OBS_COL     = 'gwl_observed'

N_DISTRICTS  = df['district_id'].nunique()
N_CATEGORIES = int(df['aquifer_category_id'].max()) + 1

# Low-station districts (Jalgaon, Akola — 1-2 wells) reported separately, never excluded
low_station_districts = sorted(df.loc[df['low_station_flag'] == 1, 'district'].unique())
print(f'Past={len(PAST_CONT)}, Future={len(FUTURE_CONT)}, Static={len(STATIC_CONT)}, '
      f'Districts={N_DISTRICTS}, Categories={N_CATEGORIES}')
print(f'Low-station districts (report separately): {low_station_districts}')
```

```python
# ── Per-fold normalisation ────────────────────────────────────────────────────
# gwl_observed is a 0/1 flag, not a magnitude — it must NOT go through the
# StandardScaler (that would turn it into a z-scored value and silently break
# the window-quality gate, which needs the raw fraction of observed days).
SCALED_PAST_CONT = [c for c in PAST_CONT if c != OBS_COL]

# gwl-like columns get the per-district target mu/sd (same scale as the target
# itself, so gwl_mean/gwl_lag_*/gwl_roll_mean_* and the target are all
# comparable to the model). gwl_diff columns (changes, std) are scaled by sd only.
GWL_LIKE_COLS = ['gwl_mean','gwl_min','gwl_max','gwl_lag_7','gwl_lag_14','gwl_lag_30',
                 'gwl_roll_mean_30','gwl_roll_mean_60','gwl_roll_mean_90']
GWL_DIFF_COLS = ['gwl_7d_change','gwl_30d_change','gwl_roll_std_30','gwl_roll_std_60','gwl_roll_std_90']

def normalize_for_fold(df, train_end):
    """Fit scalers on data up to (and including) train_end only, then transform
    the full dataframe. Returns (df_norm, tgt_stats). tgt_stats is per-district
    mean/std of the RAW target computed from train rows only — never test data."""
    train_mask = df['date'] <= pd.Timestamp(train_end)

    feature_cols = SCALED_PAST_CONT + STATIC_CONT + FUTURE_CONT
    feature_scaler = StandardScaler()
    feature_scaler.fit(df.loc[train_mask, feature_cols])
    df_norm = df.copy()
    df_norm[feature_cols] = feature_scaler.transform(df[feature_cols])

    tgt_stats = (df.loc[train_mask].groupby('district')[TARGET].agg(['mean', 'std'])
                 .rename(columns={'mean': 'mu', 'std': 'sd'}))
    tgt_stats['sd'] = tgt_stats['sd'].replace(0, 1).fillna(1)

    for c in GWL_LIKE_COLS + GWL_DIFF_COLS:
        df_norm[c] = df[c]
    dist_mu = df_norm['district'].map(tgt_stats['mu'])
    dist_sd = df_norm['district'].map(tgt_stats['sd'])
    for c in GWL_LIKE_COLS:
        df_norm[c] = (df_norm[c] - dist_mu) / dist_sd
    for c in GWL_DIFF_COLS:
        df_norm[c] = df_norm[c] / dist_sd
    # Target: same per-district z-score; NaN on unobserved days propagates
    # naturally through the arithmetic and is handled by the mask downstream.
    df_norm[TARGET] = (df[TARGET] - dist_mu) / dist_sd

    return df_norm, tgt_stats

print('normalize_for_fold() defined.')
```

```python
# ── Dataset: calendar-true windows + observed-fraction gating + target mask ───
class GWLDataset(Dataset):
    """
    Builds (encoder_len + decoder_len)-day windows. Because master_dataset.csv
    is a full daily calendar per district (no dropped rows), every window is
    genuinely `encoder_len + decoder_len` consecutive days — no silent gap
    corruption like v1/v2.

    A window is KEPT only if:
      - its encoder slice has >= enc_obs_min fraction of real (gwl_observed=1) days
      - its decoder slice has >= dec_obs_min fraction of real days
    Target values on unobserved decoder days are zero-filled (never used as
    ground truth) — the accompanying `mask` tensor (from gwl_observed) tells
    the loss/eval which positions are real. This replaces v2's behaviour of
    silently dropping any window containing a single NaN target day.
    """
    def __init__(self, df, encoder_len, decoder_len, past_cols, future_cols,
                 static_cat_cols, static_cont_cols, target_col, obs_col,
                 split_start=None, split_end=None,
                 enc_obs_min=0.70, dec_obs_min=0.50):
        self.windows = []
        self.n_dropped_enc = self.n_dropped_dec = 0
        for dist, grp in df.groupby('district'):
            grp = grp.sort_values('date').reset_index(drop=True)
            n = len(grp)
            obs = grp[obs_col].values
            for i in range(encoder_len, n - decoder_len + 1):
                target_end = grp.loc[i + decoder_len - 1, 'date']
                if split_start is not None and target_end < pd.Timestamp(split_start): continue
                if split_end   is not None and target_end > pd.Timestamp(split_end):   continue

                enc_frac = obs[i - encoder_len:i].mean()
                dec_frac = obs[i:i + decoder_len].mean()
                if enc_frac < enc_obs_min:
                    self.n_dropped_enc += 1; continue
                if dec_frac < dec_obs_min:
                    self.n_dropped_dec += 1; continue

                pc = grp.iloc[i - encoder_len:i][past_cols].values.astype(np.float32)
                fc = grp.iloc[i:i + decoder_len][future_cols].values.astype(np.float32)
                sc = grp.loc[0, static_cat_cols].values.astype(np.int64)
                sx = grp.loc[0, static_cont_cols].values.astype(np.float32)
                tg = grp.iloc[i:i + decoder_len][target_col].values.astype(np.float32)
                mk = grp.iloc[i:i + decoder_len][obs_col].values.astype(np.float32)

                if np.isnan(pc).any() or np.isnan(fc).any():
                    continue  # should not happen (verified NaN-free) — defensive skip
                tg = np.nan_to_num(tg, nan=0.0)  # masked out wherever mk==0

                self.windows.append({'past_cont': pc, 'future_cont': fc, 'static_cat': sc,
                                      'static_cont': sx, 'target': tg, 'mask': mk,
                                      'district': dist, 'date': grp.iloc[i + decoder_len - 1]['date']})

    def __len__(self): return len(self.windows)

    def __getitem__(self, idx):
        w = self.windows[idx]
        return (torch.from_numpy(w['static_cat']), torch.from_numpy(w['static_cont']),
                torch.from_numpy(w['past_cont']),  torch.from_numpy(w['future_cont']),
                torch.from_numpy(w['target']),     torch.from_numpy(w['mask']))


def build_sampler(dataset, cap=SAMPLE_WEIGHT_CAP):
    """Inverse per-district window-count sample weights, capped at `cap`x the
    minimum weight, so high-coverage districts (Nagpur, ~550 windows) don't
    drown out low-coverage ones (Dhule, ~170 windows) during training."""
    from collections import Counter
    counts = Counter(w['district'] for w in dataset.windows)
    raw_weights = {d: 1.0 / c for d, c in counts.items()}
    min_w = min(raw_weights.values())
    capped = {d: min(w, min_w * cap) for d, w in raw_weights.items()}
    sample_weights = [capped[w['district']] for w in dataset.windows]
    return WeightedRandomSampler(sample_weights, num_samples=len(dataset), replacement=True)


class MaskedWeightedQuantileLoss(nn.Module):
    """Pinball loss averaged only over positions where mask==1 (real readings).
    Unobserved decoder days contribute zero gradient instead of being trained
    against a fabricated value."""
    def __init__(self, quantiles, q50_boost=1.0):
        super().__init__()
        self.quantiles = quantiles
        w = [q50_boost if abs(q - 0.5) < 1e-6 else 1.0 for q in quantiles]
        self.register_buffer('w', torch.tensor(w).float())

    def forward(self, preds, target, mask):
        target = target.unsqueeze(-1)
        diffs  = target - preds
        denom  = mask.sum().clamp(min=1.0)
        losses = []
        for i, q in enumerate(self.quantiles):
            d = diffs[..., i]
            raw = torch.max(q * d, (q - 1) * d)
            losses.append(self.w[i] * (raw * mask).sum() / denom)
        return torch.stack(losses).sum()

print('GWLDataset, build_sampler, MaskedWeightedQuantileLoss defined.')
```

```python
# ── Model factory + per-fold training loop ────────────────────────────────────
def build_model():
    return TemporalFusionTransformer(
        n_static_cat=len(STATIC_CAT), cat_vocab_sizes=[N_DISTRICTS, N_CATEGORIES], cat_embed_dim=CAT_EMBED_DIM,
        n_static_cont=len(STATIC_CONT), n_past=len(PAST_CONT), n_future=len(FUTURE_CONT),
        encoder_len=ENCODER_LEN, decoder_len=DECODER_LEN,
        d_model=D_MODEL, n_heads=N_HEADS, lstm_layers=LSTM_LAYERS,
        dropout=DROPOUT, quantiles=QUANTILES,
    ).to(DEVICE)


def run_epoch(loader, model, criterion, optimizer=None, scheduler=None):
    training = optimizer is not None
    model.train(training)
    total, n = 0.0, 0
    with torch.set_grad_enabled(training):
        for batch in loader:
            sc, sx, px, fx, y, mk = [b.to(DEVICE) for b in batch]
            preds, _, _ = model(sc, sx, px, fx)
            loss = criterion(preds, y, mk)
            if training:
                optimizer.zero_grad(); loss.backward()
                nn.utils.clip_grad_norm_(model.parameters(), GRAD_CLIP)
                optimizer.step()
                if scheduler is not None: scheduler.step()
            total += loss.item(); n += 1
    return total / max(n, 1)


def train_one_fold(fold_name, train_ds, val_ds, max_epochs=MAX_EPOCHS, patience=PATIENCE, verbose=True):
    model = build_model()
    criterion = MaskedWeightedQuantileLoss(QUANTILES, q50_boost=Q50_LOSS_BOOST).to(DEVICE)
    optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=1e-4)

    sampler = build_sampler(train_ds)
    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, sampler=sampler,
                               num_workers=2 if DEVICE.type == 'cuda' else 0,
                               pin_memory=(DEVICE.type == 'cuda'))
    val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False,
                             num_workers=2 if DEVICE.type == 'cuda' else 0,
                             pin_memory=(DEVICE.type == 'cuda'))

    steps_per_epoch = max(len(train_loader), 1)
    total_steps  = max_epochs * steps_per_epoch
    warmup_steps = WARMUP_EPOCHS * steps_per_epoch
    def lr_lambda(step):
        if step < warmup_steps:
            return step / max(1, warmup_steps)
        p = (step - warmup_steps) / max(1, total_steps - warmup_steps)
        return 0.5 * (1.0 + math.cos(math.pi * min(p, 1.0)))
    scheduler = torch.optim.lr_scheduler.LambdaLR(optimizer, lr_lambda)

    train_losses, val_losses = [], []
    best_val, patience_ctr, best_state = float('inf'), 0, None

    if verbose:
        print(f'[{fold_name}] train={len(train_ds):,} windows | val={len(val_ds):,} windows | '
              f'{steps_per_epoch} batches/epoch | patience={patience}')

    for epoch in range(1, max_epochs + 1):
        t0 = time.time()
        tr = run_epoch(train_loader, model, criterion, optimizer, scheduler)
        val = run_epoch(val_loader, model, criterion)
        train_losses.append(tr); val_losses.append(val)
        if verbose:
            print(f'  [{fold_name}] Ep {epoch:03d} | train={tr:.4f} val={val:.4f} | '
                  f'lr={optimizer.param_groups[0]["lr"]:.1e} | {time.time()-t0:.1f}s')
        if val < best_val:
            best_val = val; patience_ctr = 0
            best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}
        else:
            patience_ctr += 1
            if patience_ctr >= patience:
                if verbose: print(f'  [{fold_name}] Early stop (best val={best_val:.4f})')
                break

    model.load_state_dict(best_state)
    torch.save(best_state, OUT_DIR / f'tft_checkpoint_{fold_name}.pt')
    return model, train_losses, val_losses

print('build_model, train_one_fold defined.')
```

```python
# ── Mask-aware evaluation ──────────────────────────────────────────────────────
def evaluate(loader, model, eval_steps, dataset, tgt_stats):
    """Only scores positions where mask==1 (a real CGWB reading exists) — an
    unobserved day is neither trained on nor evaluated on."""
    model.eval(); records = []
    with torch.no_grad():
        for bi, batch in enumerate(loader):
            sc, sx, px, fx, y, mk = [b.to(DEVICE) for b in batch]
            preds, _, _ = model(sc, sx, px, fx)
            B = y.shape[0]
            for step in eval_steps:
                h = step + 1
                for i in range(B):
                    if mk[i, step].item() == 0:
                        continue
                    w = dataset.windows[bi*BATCH_SIZE+i]
                    mu = tgt_stats.loc[w['district'],'mu']
                    sd = tgt_stats.loc[w['district'],'sd']
                    act = float(y[i,step].cpu().numpy())            * sd + mu
                    q10 = float(preds[i,step,0].cpu().numpy())      * sd + mu
                    q50 = float(preds[i,step,1].cpu().numpy())      * sd + mu
                    q90 = float(preds[i,step,2].cpu().numpy())      * sd + mu
                    records.append({'district':w['district'],'date':str(w['date']),'horizon':f't+{h}',
                                    'actual':act,'q10':q10,'q50':q50,'q90':q90})
    return pd.DataFrame(records)


def metrics(df_pred, group=None):
    keys = ['horizon'] if group is None else [group,'horizon']
    rows = []
    for k, g in df_pred.groupby(keys):
        a, p = g['actual'].values, g['q50'].values
        covered = ((g['actual'] >= g['q10']) & (g['actual'] <= g['q90'])).mean()
        interval_width = (g['q90'] - g['q10']).mean()
        row = {'RMSE (m)':round(np.sqrt(np.mean((a-p)**2)),4),
               'MAE (m)':round(np.mean(np.abs(a-p)),4),
               'MAPE (%)':round(np.mean(np.abs((a-p)/(np.abs(a)+1e-8)))*100,2),
               'Coverage (%)':round(covered*100,1),
               'Interval (m)':round(interval_width,3),
               'n':len(g)}
        if group is None: row['Horizon'] = k
        else:             row[group], row['Horizon'] = k[0], k[1]
        rows.append(row)
    return pd.DataFrame(rows)

print('evaluate, metrics defined.')
```

```python
# ── WALK-FORWARD DRIVER — trains + evaluates all 3 folds ──────────────────────
from dateutil.relativedelta import relativedelta

all_test_preds = []
fold_models, fold_tgt_stats, fold_curves = {}, {}, {}

for fold in FOLDS:
    fold_name = fold['name']
    test_start, test_end = fold['test_start'], fold['test_end']
    train_end = pd.Timestamp(test_start) - pd.Timedelta(days=1)
    val_start = train_end - relativedelta(months=VAL_MONTHS) + pd.Timedelta(days=1)
    actual_train_end = val_start - pd.Timedelta(days=1)

    print(f'\n{"="*70}\nFOLD: {fold_name}')
    print(f'  train : ... -> {actual_train_end.date()}')
    print(f'  val   : {val_start.date()} -> {train_end.date()}  (early stopping only)')
    print(f'  test  : {test_start} -> {test_end}  (never seen during training/early stopping)')
    print('='*70)

    df_norm, tgt_stats = normalize_for_fold(df, train_end)

    train_ds = GWLDataset(df_norm, ENCODER_LEN, DECODER_LEN, PAST_CONT, FUTURE_CONT,
                          STATIC_CAT, STATIC_CONT, TARGET, OBS_COL,
                          split_start=None, split_end=str(actual_train_end.date()),
                          enc_obs_min=ENC_OBS_MIN, dec_obs_min=DEC_OBS_MIN)
    val_ds = GWLDataset(df_norm, ENCODER_LEN, DECODER_LEN, PAST_CONT, FUTURE_CONT,
                        STATIC_CAT, STATIC_CONT, TARGET, OBS_COL,
                        split_start=str(val_start.date()), split_end=str(train_end.date()),
                        enc_obs_min=ENC_OBS_MIN, dec_obs_min=DEC_OBS_MIN)
    test_ds = GWLDataset(df_norm, ENCODER_LEN, DECODER_LEN, PAST_CONT, FUTURE_CONT,
                        STATIC_CAT, STATIC_CONT, TARGET, OBS_COL,
                        split_start=test_start, split_end=test_end,
                        enc_obs_min=ENC_OBS_MIN, dec_obs_min=DEC_OBS_MIN)
    print(f'  windows: train={len(train_ds):,}  val={len(val_ds):,}  test={len(test_ds):,}')

    model, tr_losses, va_losses = train_one_fold(fold_name, train_ds, val_ds)
    fold_models[fold_name] = model
    fold_tgt_stats[fold_name] = tgt_stats
    fold_curves[fold_name] = (tr_losses, va_losses)

    test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False)
    preds = evaluate(test_loader, model, EVAL_STEPS, test_ds, tgt_stats)
    preds['fold'] = fold_name
    all_test_preds.append(preds)

    fold_metrics = metrics(preds).set_index('Horizon').sort_index()
    print(f'\n[{fold_name}] test metrics:')
    print(fold_metrics.to_string())

all_test_preds = pd.concat(all_test_preds, ignore_index=True)
all_test_preds.to_csv(OUT_DIR / 'tft_predictions_all_folds.csv', index=False)
print(f'\nSaved -> outputs/tft_predictions_all_folds.csv ({len(all_test_preds):,} rows)')
```

```python
# ── AGGREGATE RESULTS — headline is median across fold x district ────────────
# Low-station districts (Jalgaon, Akola) are excluded from the headline and
# reported in their own table — their "district mean" is 1-2 wells, not a
# district signal, and including them would distort the summary either way.
district_flags = df.drop_duplicates('district').set_index('district')['low_station_flag']
all_test_preds['low_station'] = all_test_preds['district'].map(district_flags)

main_preds = all_test_preds[all_test_preds['low_station'] == 0]
low_preds  = all_test_preds[all_test_preds['low_station'] == 1]

print('=== HEADLINE: median across (fold x district), low-station districts excluded ===')
per_dist_fold = metrics(main_preds, group='district')
headline = per_dist_fold.groupby('Horizon')[['RMSE (m)','MAE (m)','MAPE (%)','Coverage (%)']].median()
print(headline.to_string())

print('\n=== Per-season (fold) aggregate — main districts only ===')
per_season = metrics(main_preds.assign(**{'_g': main_preds['fold']}), group='_g')
per_season = per_season.rename(columns={'_g': 'season'})
print(per_season.sort_values(['Horizon','season']).to_string(index=False))

if len(low_preds):
    print('\n=== Low-station districts (reported separately, NOT in headline) ===')
    low_metrics = metrics(low_preds, group='district')
    print(low_metrics.sort_values(['Horizon','district']).to_string(index=False))

headline.to_csv(OUT_DIR / 'tft_metrics_headline.csv')
per_season.to_csv(OUT_DIR / 'tft_metrics_per_season.csv', index=False)
```

```python
# ── PER-DISTRICT METRICS (pooled across all 3 test folds) ────────────────────
per_dist = metrics(main_preds, group='district')
per_dist = per_dist[['district','Horizon','RMSE (m)','MAE (m)','MAPE (%)','Coverage (%)','Interval (m)','n']]
per_dist = per_dist.sort_values(['Horizon','MAPE (%)']).reset_index(drop=True)

print('\n=== PER-DISTRICT RESULTS (pooled monsoon + post-monsoon + dry test windows) ===\n')
for h in ['t+7','t+14','t+30']:
    sub = per_dist[per_dist['Horizon']==h].drop(columns='Horizon').reset_index(drop=True)
    print(f'--- {h} ---')
    print(sub.to_string(index=False))
    print(f'   MEAN  : RMSE={sub["RMSE (m)"].mean():.3f}  MAE={sub["MAE (m)"].mean():.3f}  MAPE={sub["MAPE (%)"].mean():.2f}%  Coverage={sub["Coverage (%)"].mean():.1f}%')
    print(f'   MEDIAN: RMSE={sub["RMSE (m)"].median():.3f}  MAE={sub["MAE (m)"].median():.3f}  MAPE={sub["MAPE (%)"].median():.2f}%  Coverage={sub["Coverage (%)"].median():.1f}%')
    print()

per_dist.to_csv(OUT_DIR/'tft_metrics_per_district.csv', index=False)
print('Saved -> outputs/tft_metrics_per_district.csv')
```

```python
# ── COVERAGE RATE ANALYSIS (pooled test predictions, all folds) ──────────────
all_test_preds['covered'] = (all_test_preds['actual'] >= all_test_preds['q10']) & (all_test_preds['actual'] <= all_test_preds['q90'])
all_test_preds['interval_width'] = all_test_preds['q90'] - all_test_preds['q10']

print('=== COVERAGE RATE ANALYSIS (genuine held-out test windows) ===')
print('Target: ~80% of actuals should fall inside the [q10, q90] interval\n')

agg_cov = all_test_preds.groupby('horizon').agg(
    coverage=('covered', lambda x: round(x.mean()*100,1)),
    mean_interval_width=('interval_width', lambda x: round(x.mean(),3)),
    n=('covered','count')
).reset_index()
print('Aggregate coverage by horizon:')
print(agg_cov.to_string(index=False))

cov_pivot = (all_test_preds[all_test_preds['low_station']==0]
             .groupby(['district','horizon'])['covered']
             .mean().mul(100).round(1).unstack('horizon'))
cov_pivot.columns.name = None
print('\nPer-district coverage (%) — target ~80 in each cell (main districts only):')
print(cov_pivot.to_string())

for h in cov_pivot.columns:
    undercovered = cov_pivot[cov_pivot[h] < 65].index.tolist()
    overcovered  = cov_pivot[cov_pivot[h] > 95].index.tolist()
    if undercovered: print(f'\n  {h} UNDERCOVERED (<65%): {undercovered}')
    if overcovered:  print(f'\n  {h} OVERCOVERED (>95%): {overcovered}')

cov_pivot.to_csv(OUT_DIR/'tft_coverage_by_district.csv')

fig, ax = plt.subplots(figsize=(6, max(6, len(cov_pivot)*0.35)))
im = ax.imshow(cov_pivot.values, aspect='auto', vmin=50, vmax=100, cmap='RdYlGn', interpolation='nearest')
ax.set_xticks(range(len(cov_pivot.columns))); ax.set_xticklabels(cov_pivot.columns)
ax.set_yticks(range(len(cov_pivot.index))); ax.set_yticklabels(cov_pivot.index, fontsize=8)
ax.set_title('80% Interval Coverage Rate (%) per District\n(pooled monsoon+post-monsoon+dry test)')
plt.colorbar(im, ax=ax, label='Coverage (%)')
for i in range(len(cov_pivot.index)):
    for j in range(len(cov_pivot.columns)):
        v = cov_pivot.values[i, j]
        ax.text(j, i, f'{v:.0f}', ha='center', va='center', fontsize=7,
                color='black' if 65 < v < 95 else 'white')
plt.tight_layout(); plt.savefig(OUT_DIR/'tft_coverage_heatmap.png', dpi=150); plt.show()
```

```python
# ── CONFORMAL CALIBRATION (per-fold, no test leakage) ─────────────────────────
# Margins are computed independently within each fold's own test predictions,
# not pooled across folds — a monsoon-fold margin should not calibrate a
# dry-season fold's intervals, since the seasons have different residual behaviour.
TARGET_COVERAGE = 0.80
cal_parts = []
print('=== CONFORMAL CALIBRATION MARGINS (per fold) ===')
for fold in FOLDS:
    fname = fold['name']
    g_fold = all_test_preds[all_test_preds['fold'] == fname].copy()
    g_fold['nc_score'] = np.maximum(g_fold['q10'] - g_fold['actual'], g_fold['actual'] - g_fold['q90'])
    for h, g in g_fold.groupby('horizon'):
        n = len(g)
        level = min(np.ceil((n + 1) * TARGET_COVERAGE) / n, 1.0)
        margin = float(np.quantile(g['nc_score'], level))
        mask = (g_fold['horizon'] == h)
        g_fold.loc[mask, 'q10_cal'] = g_fold.loc[mask, 'q10'] - margin
        g_fold.loc[mask, 'q90_cal'] = g_fold.loc[mask, 'q90'] + margin
        print(f'  {fname} {h}: margin = {margin:+.4f} m')
    cal_parts.append(g_fold)

cal_preds = pd.concat(cal_parts, ignore_index=True)
cal_preds['covered_cal'] = (cal_preds['actual'] >= cal_preds['q10_cal']) & (cal_preds['actual'] <= cal_preds['q90_cal'])
cal_preds['width_cal'] = cal_preds['q90_cal'] - cal_preds['q10_cal']

print('\nCoverage after calibration (by horizon, pooled across folds):')
check = cal_preds.groupby('horizon').agg(
    before=('covered', lambda x: round(x.mean()*100,1)),
    after=('covered_cal', lambda x: round(x.mean()*100,1)),
    mean_width_before=('interval_width', lambda x: round(x.mean(),3)),
    mean_width_after=('width_cal', lambda x: round(x.mean(),3)),
).reset_index()
print(check.to_string(index=False))

cal_preds.to_csv(OUT_DIR/'tft_predictions_calibrated.csv', index=False)
print('\nSaved -> outputs/tft_predictions_calibrated.csv')
```

```python
# ── VSN feature importance (fold3/dry-season model — most data, most recent) ─
model_final.eval()
df_norm_final, tgt_stats_final = normalize_for_fold(df, pd.Timestamp(FOLDS[-1]['test_start']) - pd.Timedelta(days=1))
imp_ds = GWLDataset(df_norm_final, ENCODER_LEN, DECODER_LEN, PAST_CONT, FUTURE_CONT, STATIC_CAT, STATIC_CONT,
                    TARGET, OBS_COL, split_start=FOLDS[-1]['test_start'], split_end=FOLDS[-1]['test_end'],
                    enc_obs_min=ENC_OBS_MIN, dec_obs_min=DEC_OBS_MIN)
imp_loader = DataLoader(imp_ds, batch_size=BATCH_SIZE, shuffle=False)

pw, fw, sw = [], [], []
with torch.no_grad():
    for i, batch in enumerate(imp_loader):
        if i >= 30: break
        sc, sx, px, fx, _, _ = [b.to(DEVICE) for b in batch]
        _, vsn_w, _ = model_final(sc, sx, px, fx)
        pw.append(vsn_w['past'].cpu().mean(dim=(0,1)))
        fw.append(vsn_w['future'].cpu().mean(dim=(0,1)))
        sw.append(vsn_w['static'].cpu().mean(dim=0))
past_imp = torch.stack(pw).mean(0).numpy()
fut_imp  = torch.stack(fw).mean(0).numpy()
stat_imp = torch.stack(sw).mean(0).numpy()

static_labels = [f'{c}_emb' for c in STATIC_CAT] + STATIC_CONT

fig, axes = plt.subplots(1,3,figsize=(20,8))
for ax, imp, names, color, title in [
    (axes[0], past_imp, PAST_CONT, 'steelblue', 'Past Features (VSN)'),
    (axes[1], fut_imp,  FUTURE_CONT,'coral',    'Future Features (VSN)'),
    (axes[2], stat_imp, static_labels, 'seagreen', 'Static Features (VSN)'),
]:
    idx = np.argsort(imp)[::-1]
    ax.barh([names[i] for i in idx], imp[idx], color=color)
    ax.set_title(title); ax.set_xlabel('Mean VSN weight'); ax.invert_yaxis()
plt.tight_layout()
plt.savefig(OUT_DIR/'tft_vsn_importance.png', dpi=150); plt.show()
```
